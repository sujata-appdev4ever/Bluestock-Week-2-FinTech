```python
import requests
import pandas as pd
```

```python
url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

print(response.status_code)
```

```python
data = response.json()

print(type(data))
print(data[:3])
```

```python
df = pd.DataFrame(data)

df.head()
```

```python
print(df.shape)
print(df.columns)
print(df.dtypes)
```

```python
df.to_csv("api_data.csv", index=False)

print("CSV file created successfully!")
```
